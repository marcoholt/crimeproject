### Feature Engineering

```python
#create a monthly crime count dataframe
monthly_crime = (
    df.groupby(['area_name', 'year', 'month'])
      .size()
      .reset_index(name='crime_count')
)
```

```python
#previous crime count
monthly_crime = monthly_crime.sort_values(
    ['area_name', 'year', 'month']
)

monthly_crime['previous_crime_count'] = (
    monthly_crime.groupby('area_name')['crime_count']
                 .shift(1)
)
```

```python
#create rolling average 
monthly_crime['rolling_average'] = (
    monthly_crime.groupby('area_name')['crime_count']
                 .transform(
                     lambda x: x.shift(1)
                               .rolling(window=3)
                               .mean()
                 )
)
```

```python
#crime trend
monthly_crime['crime_trend'] = (
    monthly_crime['crime_count']
    - monthly_crime['previous_crime_count']
)
```

```python
#check the new columns
monthly_crime.head(50)
```

Adding more features for better generalization

```python
# Make sure data is sorted correctly by neighborhood and time
monthly_crime = monthly_crime.sort_values(['area_name', 'year', 'month']).copy()

# -----------------------------
# Additional Lag Features
# -----------------------------
# Previous 1 month
monthly_crime['lag_1'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .shift(1)
)

# Previous 3 months
monthly_crime['lag_3'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .shift(3)
)

# Previous 6 months
monthly_crime['lag_6'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .shift(6)
)

# Previous 12 months
monthly_crime['lag_12'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .shift(12)
)

# -----------------------------
# Additional Rolling Averages
# -----------------------------
# 3-month rolling average
monthly_crime['rolling_3'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .transform(lambda x: x.shift(1).rolling(window=3).mean())
)

# 6-month rolling average
monthly_crime['rolling_6'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .transform(lambda x: x.shift(1).rolling(window=6).mean())
)

# 12-month rolling average
monthly_crime['rolling_12'] = (
    monthly_crime
    .groupby('area_name')['crime_count']
    .transform(lambda x: x.shift(1).rolling(window=12).mean())
)

# -----------------------------
# Cyclical Month Features
# -----------------------------
# Helps model understand that December and January are close together
monthly_crime['month_sin'] = np.sin(2 * np.pi * monthly_crime['month'] / 12)
monthly_crime['month_cos'] = np.cos(2 * np.pi * monthly_crime['month'] / 12)

# -----------------------------
# Optional: Drop rows with missing lag/rolling values
# This removes early months where lag_12 or rolling_12 cannot be calculated
# -----------------------------
monthly_crime_model = monthly_crime.dropna().copy()

# Check new features
monthly_crime_model.head()
```

### Engineered Features

### Crime Count (`crime_count`)
The `crime_count` feature measures the total number of crimes occurring in a neighborhood during a given month. This serves as the target variable for the forecasting model and represents the value the model is trying to predict.

### Previous Crime Count (`previous_crime_count`)
The `previous_crime_count` feature captures the number of crimes that occurred in the neighborhood during the previous month. This provides historical context and helps the model learn patterns from recent crime activity.

### Rolling Average (`rolling_average`)
The `rolling_average` feature calculates the average crime count over the previous three months. This helps smooth short-term fluctuations and allows the model to recognize longer-term crime patterns and trends.

### Crime Trend (`crime_trend`)
The `crime_trend` feature measures the month-to-month change in crime activity by comparing the current month's crime count to the previous month's crime count. This helps the model identify whether crime levels are increasing, decreasing, or remaining stable over time.

### Additional Feature Engineering

To improve forecasting performance and help the model generalize to future crime patterns, several additional time-based features were engineered. These features provide historical context, capture seasonal behavior, and help identify short-term and long-term crime trends within each neighborhood.

- **lag_1** captures the crime count from the previous month and provides immediate historical context for forecasting future crime activity.

- **lag_3** captures crime activity from three months prior and helps the model identify medium-term neighborhood crime patterns.

- **lag_6** captures crime activity from six months prior and provides additional insight into longer-term crime behavior within a neighborhood.

- **lag_12** captures crime activity from the same month in the previous year, helping the model learn yearly seasonal patterns and recurring crime trends.

- **rolling_3** calculates the average crime count over the previous three months, smoothing short-term fluctuations and highlighting recent crime trends.

- **rolling_6** calculates the average crime count over the previous six months, helping identify medium-term changes in neighborhood crime activity.

- **rolling_12** calculates the average crime count over the previous twelve months, providing a long-term view of crime patterns and reducing the impact of temporary spikes or declines.

- **month_sin** and **month_cos** transform the month variable into cyclical features, allowing the model to understand that months repeat annually and that December and January are temporally adjacent rather than far apart numerically.

Overall, these engineered features were created to help the model capture temporal dependencies, seasonality, and neighborhood-specific crime trends more effectively. By incorporating both short-term and long-term historical information, the model is better equipped to generalize to future crime patterns and improve forecasting performance.

## Model Development

##### ML Forecasting

#### Final Forecasting Results (2024 Holdout Year)

```python
# Train on historical years
train_data = monthly_crime_model[
    monthly_crime_model['year'] < 2024
]

# Test on future year
test_data = monthly_crime_model[
    monthly_crime_model['year'] == 2024
]

# Features and target
X_train = train_data.drop('crime_count', axis=1)
y_train = train_data['crime_count']

X_test = test_data.drop('crime_count', axis=1)
y_test = test_data['crime_count']

print("Training Rows:", len(X_train))
print("Testing Rows:", len(X_test))
```

```python
numeric_features = [
    'previous_crime_count',
    'rolling_average',
    'month'
]

categorical_features = [
    'area_name'
]
```

```python


# Numeric preprocessing
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

##Baseline Models
# Linear Regression Pipeline
linear_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

# Random Forest Pipeline
rf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(random_state=42))
])

# Train models
linear_pipeline.fit(X_train, y_train)
rf_pipeline.fit(X_train, y_train)
```

Training RandomForest and LinearRegression

```python
# Predictions
linear_preds = linear_pipeline.predict(X_test)
rf_preds = rf_pipeline.predict(X_test)

# Evaluation function
def evaluate_model(model_name, y_true, predictions):

    mae = mean_absolute_error(y_true, predictions)
    mse = mean_squared_error(y_true, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, predictions)

    return {
        'Model': model_name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2': r2
    }

# Store results
results = [
    evaluate_model(
        'Linear Regression',
        y_test,
        linear_preds
    ),
    evaluate_model(
        'Random Forest Regressor',
        y_test,
        rf_preds
    )
]

# Results table
results_df = pd.DataFrame(results)

results_df
```

Training XGBoost

```python
from xgboost import XGBRegressor

# XGBoost Pipeline
xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', XGBRegressor(
        objective='reg:squarederror',
        random_state=42
    ))
])

# Train XGBoost model
xgb_pipeline.fit(X_train, y_train)

# Make predictions
xgb_preds = xgb_pipeline.predict(X_test)

# Add XGBoost results
results.append(
    evaluate_model(
        'XGBoost Regressor',
        y_test,
        xgb_preds
    )
)

# Updated results table
results_df = pd.DataFrame(results)
results_df
```

```python
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

# Time Series Cross Validation
tscv = TimeSeriesSplit(n_splits=5)

# -----------------------------
# RIDGE REGRESSION
# -----------------------------
ridge_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', Ridge())
])

ridge_params = {
    'model__alpha': [0.01, 0.1, 1, 10, 100]
}

ridge_search = GridSearchCV(
    ridge_pipeline,
    ridge_params,
    cv=tscv,
    scoring='r2',
    n_jobs=-1
)

ridge_search.fit(X_train, y_train)

# -----------------------------
# RANDOM FOREST
# -----------------------------
rf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(random_state=42))
])

rf_params = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [5, 10, None],
    'model__min_samples_split': [2, 5, 10],
    'model__min_samples_leaf': [1, 2, 4]
}

rf_search = GridSearchCV(
    rf_pipeline,
    rf_params,
    cv=tscv,
    scoring='r2',
    n_jobs=-1
)

rf_search.fit(X_train, y_train)

# -----------------------------
# XGBOOST
# -----------------------------
xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', XGBRegressor(
        objective='reg:squarederror',
        random_state=42
    ))
])

xgb_params = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [3, 5],
    'model__learning_rate': [0.01, 0.05, 0.1],
    'model__subsample': [0.8, 1.0]
}

xgb_search = GridSearchCV(
    xgb_pipeline,
    xgb_params,
    cv=tscv,
    scoring='r2',
    n_jobs=-1
)

xgb_search.fit(X_train, y_train)

# -----------------------------
# BEST MODELS
# -----------------------------
print("Best Ridge Parameters:")
print(ridge_search.best_params_)
print("Best Ridge CV R²:", ridge_search.best_score_)

print("\nBest Random Forest Parameters:")
print(rf_search.best_params_)
print("Best Random Forest CV R²:", rf_search.best_score_)

print("\nBest XGBoost Parameters:")
print(xgb_search.best_params_)
print("Best XGBoost CV R²:", xgb_search.best_score_)

# -----------------------------
# TEST SET EVALUATION (2024)
# -----------------------------
best_models = {
    "Ridge Regression": ridge_search.best_estimator_,
    "Random Forest": rf_search.best_estimator_,
    "XGBoost": xgb_search.best_estimator_
}

results = []

for name, model in best_models.items():
    preds = model.predict(X_test)

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, preds),
        "MSE": mean_squared_error(y_test, preds),
        "RMSE": np.sqrt(mean_squared_error(y_test, preds)),
        "R2": r2_score(y_test, preds)
    })

tuned_results_df = pd.DataFrame(results)
tuned_results_df
```

#### SARIMA Forecasting Model



```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

sarima_results = []

# Make sure data is sorted
monthly_crime_forecast = monthly_crime.sort_values(['area_name', 'year', 'month']).copy()

for area in monthly_crime_forecast['area_name'].unique():
    
    area_data = monthly_crime_forecast[
        monthly_crime_forecast['area_name'] == area
    ].sort_values(['year', 'month'])
    
    train = area_data[area_data['year'] < 2024]['crime_count']
    test = area_data[area_data['year'] == 2024]['crime_count']
    
    if len(train) < 24 or len(test) == 0:
        continue
    
    try:
        model = SARIMAX(
            train,
            order=(1, 1, 1),
            seasonal_order=(1, 1, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False
        )
        
        model_fit = model.fit(disp=False)
        preds = model_fit.forecast(steps=len(test))
        
        sarima_results.append({
            'area_name': area,
            'MAE': mean_absolute_error(test, preds),
            'MSE': mean_squared_error(test, preds),
            'RMSE': np.sqrt(mean_squared_error(test, preds)),
            'R2': r2_score(test, preds)
        })
        
    except Exception as e:
        print(f"SARIMA failed for {area}: {e}")

sarima_results_df = pd.DataFrame(sarima_results)

sarima_summary = pd.DataFrame([{
    'Model': 'SARIMA',
    'MAE': sarima_results_df['MAE'].mean(),
    'MSE': sarima_results_df['MSE'].mean(),
    'RMSE': sarima_results_df['RMSE'].mean(),
    'R2': sarima_results_df['R2'].mean()
}])

sarima_summary
```

#### Prophet Forecasting Model

```python
# If Prophet is not installed, run this once:

from prophet import Prophet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd
import numpy as np

prophet_results = []

# Prepare forecasting dataframe
monthly_crime_forecast = monthly_crime.sort_values(['area_name', 'year', 'month']).copy()

# Create date column
monthly_crime_forecast['date'] = pd.to_datetime(
    monthly_crime_forecast[['year', 'month']].assign(day=1)
)

for area in monthly_crime_forecast['area_name'].unique():

    area_data = monthly_crime_forecast[
        monthly_crime_forecast['area_name'] == area
    ].sort_values('date').copy()

    train = area_data[area_data['year'] < 2024]
    test = area_data[area_data['year'] == 2024]

    if len(train) < 24 or len(test) == 0:
        continue

    prophet_train = train[['date', 'crime_count']].rename(
        columns={
            'date': 'ds',
            'crime_count': 'y'
        }
    )

    try:
        model = Prophet(
            yearly_seasonality=True,
            weekly_seasonality=False,
            daily_seasonality=False
        )

        model.fit(prophet_train)

        future = test[['date']].rename(columns={'date': 'ds'})
        forecast = model.predict(future)

        preds = forecast['yhat'].values
        actuals = test['crime_count'].values

        prophet_results.append({
            'area_name': area,
            'MAE': mean_absolute_error(actuals, preds),
            'MSE': mean_squared_error(actuals, preds),
            'RMSE': np.sqrt(mean_squared_error(actuals, preds)),
            'R2': r2_score(actuals, preds)
        })

    except Exception as e:
        print(f"Prophet failed for {area}: {e}")

prophet_results_df = pd.DataFrame(prophet_results)

prophet_summary = pd.DataFrame([{
    'Model': 'Prophet',
    'MAE': prophet_results_df['MAE'].mean(),
    'MSE': prophet_results_df['MSE'].mean(),
    'RMSE': prophet_results_df['RMSE'].mean(),
    'R2': prophet_results_df['R2'].mean()
}])

prophet_summary
```

#### Cross Validation Results (Supplementary Analysis)

```python
from sklearn.model_selection import TimeSeriesSplit, cross_val_score
import numpy as np
import pandas as pd

tscv = TimeSeriesSplit(n_splits=5)

models = {
    'Linear Regression': linear_pipeline,
    'Random Forest Regressor': rf_pipeline,
    'XGBoost Regressor': xgb_pipeline
}

cv_results = []

for model_name, model_pipeline in models.items():

    r2_scores = cross_val_score(
        model_pipeline,
        X_train,
        y_train,
        cv=tscv,
        scoring='r2'
    )

    neg_mse_scores = cross_val_score(
        model_pipeline,
        X_train,
        y_train,
        cv=tscv,
        scoring='neg_mean_squared_error'
    )

    rmse_scores = np.sqrt(-neg_mse_scores)

    cv_results.append({
        'Model': model_name,
        'CV R2 Scores': r2_scores,
        'Mean CV R2': r2_scores.mean(),
        'Std CV R2': r2_scores.std(),
        'CV RMSE Scores': rmse_scores,
        'Mean CV RMSE': rmse_scores.mean(),
        'Std CV RMSE': rmse_scores.std()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df
```

### Predict 2025 Neighborhood Crime - Investment Descion

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
import numpy as np
import pandas as pd

# -----------------------------
# Final Features
# -----------------------------
numeric_features = [
    'year', 'month',
    'lag_1', 'lag_3', 'lag_6', 'lag_12',
    'rolling_3', 'rolling_6', 'rolling_12',
    'crime_trend',
    'month_sin', 'month_cos'
]

categorical_features = ['area_name']

features = categorical_features + numeric_features

# -----------------------------
# Rebuild Preprocessor
# -----------------------------
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

forecast_preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# -----------------------------
# Train Final Ridge Model on 2020-2024
# -----------------------------
train_2020_2024 = monthly_crime_model[
    monthly_crime_model['year'] <= 2024
].copy()

X_final = train_2020_2024[features]
y_final = train_2020_2024['crime_count']

final_ridge_model = Pipeline([
    ('preprocessor', forecast_preprocessor),
    ('model', Ridge(alpha=10))
])

final_ridge_model.fit(X_final, y_final)

print("Final Ridge model trained successfully.")

# -----------------------------
# Create 2025 Forecast Dataset
# -----------------------------
future_rows = []

for area in monthly_crime_model['area_name'].unique():

    area_df = (
        monthly_crime_model[monthly_crime_model['area_name'] == area]
        .sort_values(['year', 'month'])
        .copy()
    )

    latest = area_df.iloc[-1]

    for month in range(1, 13):

        future_rows.append({
            'area_name': area,
            'year': 2025,
            'month': month,

            'lag_1': latest['lag_1'],
            'lag_3': latest['lag_3'],
            'lag_6': latest['lag_6'],
            'lag_12': latest['lag_12'],

            'rolling_3': latest['rolling_3'],
            'rolling_6': latest['rolling_6'],
            'rolling_12': latest['rolling_12'],

            'crime_trend': latest['crime_trend'],

            'month_sin': np.sin(2 * np.pi * month / 12),
            'month_cos': np.cos(2 * np.pi * month / 12)
        })

future_2025 = pd.DataFrame(future_rows)

# -----------------------------
# Predict 2025 Crime Counts
# -----------------------------
future_2025['predicted_crime_count'] = final_ridge_model.predict(
    future_2025[features]
)

future_2025['predicted_crime_count'] = (
    future_2025['predicted_crime_count'].clip(lower=0)
)

future_2025.head()
```

```python
forecast_summary = (
    future_2025
    .groupby('area_name')['predicted_crime_count']
    .agg(['mean', 'sum'])
    .reset_index()
    .rename(columns={
        'mean': 'avg_monthly_predicted_crime',
        'sum': 'total_predicted_2025_crime'
    })
    .sort_values('total_predicted_2025_crime', ascending=False)
)

forecast_summary
```
